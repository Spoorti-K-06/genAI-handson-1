```python
from transformers import pipeline
```

```python
story_generator = pipeline(
    "text-generation",
    model="gpt2"
)
```

```python
prompt = "The knight entered the dark cave"
```

```python
story = story_generator(
    prompt,
    max_length=120,
    do_sample=True,
    temperature=0.9,
    top_k=50,
    top_p=0.95
)

print(story[0]["generated_text"])
```

```python
def generate_story(starting_sentence):
    result = story_generator(
        starting_sentence,
        max_length=120,
        do_sample=True,
        temperature=0.9,
        top_k=50,
        top_p=0.95
    )
    return result[0]["generated_text"]

user_input = input("Enter a starting sentence: ")
print("\nGenerated Story:\n")
print(generate_story(user_input))
```
